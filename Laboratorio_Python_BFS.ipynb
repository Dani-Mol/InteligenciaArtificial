## 2. Diccionarios Avanzados (`defaultdict`)

Para representar un grafo en programación, la forma más eficiente suele ser una **Lista de Adyacencia**, utilizando un diccionario.

Si usamos un `dict` normal, tendríamos que verificar manualmente si la llave (el nodo) existe antes de agregarle un vecino. Al usar `defaultdict` de la librería nativa `collections`, Python crea el valor por defecto (en este caso, una lista vacía) si la llave aún no existe, ahorrándonos validaciones innecesarias.

```python
from collections import defaultdict

# Creamos un diccionario donde el valor por defecto es una lista vacía []
grafo = defaultdict(list)

# Agregamos aristas (como el grafo no es dirigido, la conexión va en ambos sentidos)
aristas = [(1, 2), (1, 3), (2, 4)]

for u, v in aristas:
    print(f'{u}, {v}')
    grafo[u].append(v)
    grafo[v].append(u)
print()

print(f"Diccionario generado: {grafo.items()}\n")


print("Lista de adyacencia del grafo:")
for nodo, vecinos in grafo.items():
    print(f"Nodo {nodo} conecta con: {vecinos}")
```

## 3. Colas Eficientes (`deque`)

El algoritmo BFS procesa los nodos nivel por nivel expandiéndose en forma de abanico, por lo que requiere usar una estructura **FIFO** (First-In, First-Out o Primeras Entradas, Primeras Salidas).

Usar una lista normal y hacer `lista.pop(0)` es ineficiente: toma tiempo lineal $\mathcal{O}(N)$ porque debe recorrer todos los elementos para reacomodarlos. En su lugar, usaremos `deque` (Double-Ended Queue), que permite extracciones al inicio de la estructura en tiempo constante $\mathcal{O}(1)$.

```python
from collections import deque

# Inicializamos la cola con un nodo de inicio
nodo_inicial = 1
cola = deque([nodo_inicial])

# Simulamos el proceso de encolado (append agrega al final)
cola.append(2)
cola.append(3)

# popleft() extrae eficientemente el primer elemento
siguiente = cola.popleft()
print(f"Procesando el nodo: {siguiente}")
print(f"Estado actual de la cola (quedan los siguientes niveles): {cola}")
```

## 4. Listas por Comprensión (List Comprehensions)

```python
# Supongamos que estas son las distancias calculadas para los nodos 1, 2, 3, 4, 5
# El índice del arreglo coincide con el nodo (el índice 0 lo ignoramos)
distancias = [0, 0, 6, 6, 12, -1]
nodo_s = 1
n = 5

# Sintaxis: [expresion for variable in iterable if condicion]
# Queremos los nodos del 1 al n (inclusive), saltándonos el nodo_s
resultado = [distancias[i] for i in range(1, n + 1) if i != nodo_s]


print(f"Distancias finales excluyendo el nodo inicial ({nodo_s}): {resultado}")
```

## 5. Algoritmo BFS

Dado un grafo con `n` nodos y una lista de aristas (`edges`), queremos encontrar la distancia más corta desde un nodo inicial `s` a todos los demás. Cada arista tiene un peso fijo de 6 (todos los saltos valen lo mismo, escenario ideal para BFS).

**El plan paso a paso:**
1. **Construcción:** Crear el grafo iterando las aristas e insertándolas en un `defaultdict(list)`. Importante: como es un grafo no dirigido, debemos agregar la conexión tanto de `u` hacia `v` como de `v` hacia `u`.
2. **Inicialización del Estado:**
   - Crear un arreglo de `distancias` rellenado con `-1` de tamaño `n + 1` (así podemos usar el número del nodo directamente como índice, sin restar 1).
   - La distancia del nodo `s` a sí mismo es `0`.
3. **La Búsqueda (Bucle `while`):**
   - Iniciar una `deque` que contenga solo a `s`.
   - Mientras la cola no esté vacía, extraer el primer elemento (lo llamaremos `nodo_actual`).
   - Iterar sobre cada `vecino` conectado a `nodo_actual`.
   - Si la distancia del vecino sigue siendo `-1` (nunca ha sido visitado), calculamos su nueva distancia (`distancia[nodo_actual]`) y lo agregamos al final de la cola para procesar sus propios vecinos más tarde.
4. **Retorno:** Devolver la lista filtrada de distancias ignorando la posición `s`, tal como vimos en la sección anterior.

```python
def bfs(n, m, edges, s):
    # 1. Construcción del grafo
    grafo = defaultdict(list)
    for u, v in edges:
        grafo[u].append(v)
        grafo[v].append(u)

    # 2. Inicialización del estado
    distancias = [-1] * (n + 1)
    distancias[s] = 0

    cola = deque([s])

    # 3. La Búsqueda (Exploración nivel por nivel)
    while cola:
        nodo_actual = cola.popleft()

        # Revisamos todos los vecinos conectados al nodo actual
        for vecino in grafo[nodo_actual]:
            # Si el vecino no ha sido descubierto (-1)
            if distancias[vecino] == -1:
                distancias[vecino] = distancias[nodo_actual] + 1
                cola.append(vecino)

    # 4. Retorno (Formateo con lista por comprensión)
    return [distancias[i] for i in range(1, n + 1) if i != s]
```
